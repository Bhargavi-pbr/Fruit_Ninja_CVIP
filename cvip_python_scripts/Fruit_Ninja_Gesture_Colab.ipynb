```python
import numpy as np
import matplotlib.pyplot as plt
import math

# Step 2: Line Segment to Circle Collision Formulation
def check_line_circle_intersection(p1, p2, circle_center, radius):
    """
    Determines if line segment (p1 -> p2) intersects a circle of radius r.
    p1, p2: (x, y) tuples
    circle_center: (cx, cy) tuple
    """
    x1, y1 = p1
    x2, y2 = p2
    cx, cy = circle_center
    
    dx = x2 - x1
    dy = y2 - y1
    len_sq = dx * dx + dy * dy
    
    if len_sq == 0:
        return math.hypot(x1 - cx, y1 - cy) <= radius
        
    # Projection scalar clamped to [0, 1]
    t = max(0.0, min(1.0, ((cx - x1) * dx + (cy - y1) * dy) / len_sq))
    closest_x = x1 + t * dx
    closest_y = y1 + t * dy
    
    dist_sq = (cx - closest_x)**2 + (cy - closest_y)**2
    return dist_sq <= radius**2

# Test case:
fruit = (200, 200)
r = 40
blade_start = (100, 250)
blade_end = (280, 160)

hit = check_line_circle_intersection(blade_start, blade_end, fruit, r)
print(f"Blade cut fruit: {hit}")
```

```python
# Step 3: Visualize the Slicing Geometry
fig, ax = plt.subplots(figsize=(6, 6))
circle = plt.Circle(fruit, r, color='crimson', alpha=0.6, label='Watermelon Target')
ax.add_patch(circle)

# Draw blade segment
ax.plot([blade_start[0], blade_end[0]], [blade_start[1], blade_end[1]], color='cyan', linewidth=3, label='Katana Slash Segment')
ax.scatter([blade_start[0], blade_end[0]], [blade_start[1], blade_end[1]], color='yellow', zorder=5)

ax.set_xlim(50, 350)
ax.set_ylim(100, 300)
ax.set_aspect('equal')
ax.set_title('CVIP Slicing Collision Geometry: Line to Circle Projection')
ax.legend()
ax.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
